### Подготовим данные (тот же DataFrame)

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Aruzhan", "Dias", "Batyr", "Amina", "Nursultan"],
    "Age":  [19, 21, 20, 22, 18],
    "Score":[78, 88, 95, 81, 67]
}
df = pd.DataFrame(data)
df
```

## Task 6: Line Plot

- Line plot scores
- Label axes
- Add title

```python
plt.plot(df["Name"], df["Score"])
plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Student Scores (Line Plot)")
plt.show()
```

## Task 7: Bar Chart

- Bar chart names vs scores

```python
plt.bar(df["Name"], df["Score"])
plt.xlabel("Student")
plt.ylabel("Score")
plt.title("Student Scores (Bar Chart)")
plt.show()
```
